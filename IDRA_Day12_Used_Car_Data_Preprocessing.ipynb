## Outlier Detection Using IQR

The Interquartile Range (IQR) method is used to identify outliers.

IQR = Q3 - Q1

Lower limit = Q1 - 1.5 × IQR

Upper limit = Q3 + 1.5 × IQR

Values below the lower limit or above the upper limit are considered outliers.

```python
numerical_columns = df.select_dtypes(include=np.number).columns.tolist()

for column in numerical_columns:
    Q1 = df[column].quantile(0.25)
    Q3 = df[column].quantile(0.75)

    IQR = Q3 - Q1

    lower_limit = Q1 - 1.5 * IQR
    upper_limit = Q3 + 1.5 * IQR

    outliers = df[(df[column] < lower_limit) | (df[column] > upper_limit)]

    print(column, ":", len(outliers), "outliers")
```

```python
for column in numerical_columns:
    Q1 = df[column].quantile(0.25)
    Q3 = df[column].quantile(0.75)

    IQR = Q3 - Q1

    lower_limit = Q1 - 1.5 * IQR
    upper_limit = Q3 + 1.5 * IQR

    df[column] = df[column].clip(lower=lower_limit, upper=upper_limit)
```

```python
for column in numerical_columns:
    Q1 = df[column].quantile(0.25)
    Q3 = df[column].quantile(0.75)

    IQR = Q3 - Q1

    lower_limit = Q1 - 1.5 * IQR
    upper_limit = Q3 + 1.5 * IQR

    outliers = df[(df[column] < lower_limit) | (df[column] > upper_limit)]

    print(column, ":", len(outliers), "outliers")
```

```python
X = df.drop('Resale_Price_Lakh', axis=1)

y = df['Resale_Price_Lakh']

print("Features shape:", X.shape)
print("Target shape:", y.shape)
```

## Train-Test Split

The dataset is divided into:
- 80% training data
- 20% testing data

The random_state is used so that the same split can be reproduced.

```python
X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42
)

print("Training data:", X_train.shape)
print("Testing data:", X_test.shape)
```

```python
categorical_columns = X_train.select_dtypes(include='object').columns.tolist()

numerical_columns = X_train.select_dtypes(include=np.number).columns.tolist()

print("Categorical Columns:")
print(categorical_columns)

print("\nNumerical Columns:")
print(numerical_columns)
```

## Encoding Categorical Variables

The categorical variables are nominal variables, so One-Hot Encoding is suitable.

One-Hot Encoding converts categories into separate numerical columns.

The encoder is fitted only on the training data to prevent data leakage.

```python
encoder = OneHotEncoder(
    handle_unknown='ignore',
    sparse_output=False
)

X_train_encoded = encoder.fit_transform(X_train[categorical_columns])

X_test_encoded = encoder.transform(X_test[categorical_columns])
```

```python
encoder.fit_transform(X_train[categorical_columns])
```

```python
encoder.transform(X_test[categorical_columns])
```

## Feature Scaling

StandardScaler is used to standardize the numerical features.

The scaler is fitted only on the training data and then used to transform both training and testing data.

This prevents data leakage.


```python
scaler = StandardScaler()

X_train_scaled = scaler.fit_transform(X_train[numerical_columns])

X_test_scaled = scaler.transform(X_test[numerical_columns])
```

```python
X_train_processed = np.hstack([
    X_train_scaled,
    X_train_encoded
])

X_test_processed = np.hstack([
    X_test_scaled,
    X_test_encoded
])

print("Processed training shape:", X_train_processed.shape)
print("Processed testing shape:", X_test_processed.shape)
```

```python
encoded_columns = encoder.get_feature_names_out(categorical_columns)

print(encoded_columns)
```

```python
processed_columns = numerical_columns + encoded_columns.tolist()

print("Number of processed columns:", len(processed_columns))
```

```python
X_train_processed_df = pd.DataFrame(
    X_train_processed,
    columns=processed_columns,
    index=X_train.index
)

X_train_processed_df.head()
```

```python
X_test_processed_df = pd.DataFrame(
    X_test_processed,
    columns=processed_columns,
    index=X_test.index
)

X_test_processed_df.head()
```

```python
train_processed = X_train_processed_df.copy()

train_processed['Resale_Price_Lakh'] = y_train

test_processed = X_test_processed_df.copy()

test_processed['Resale_Price_Lakh'] = y_test
```
